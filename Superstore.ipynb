# Cleaning Data 

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ","_")
    .str.replace("-","_")
)

df.head()
        
```

```python
# Convert date columns from text to datetime
import pandas as pd

df["order_date"] = pd.to_datetime(df["order_date"])
df["ship_date"] = pd.to_datetime(df["ship_date"])

df[["order_date", "ship_date"]].head()
```

# Creating calculated fields 

```python
df["profit_margin"] = df["profit"] / df["sales"]

df["shipping_days"] = (df["ship_date"] - df["order_date"]).dt.days

df["order_month"] = df["order_date"].dt.to_period("M").astype(str)

def discount_type(discount):
    if discount == 0:
        return "No Discount"
    elif discount <= 0.2:
        return "Low Discount"
    elif discount <= 0.5:
        return "Medium Discount"
    else :
        return "High Discount"

df["discount_type"] = df["discount"].apply(discount_type)

    
df.head() 
```

# Data Analysis

```python
overall_performance = run_query ("""

SELECT 

COUNT(DISTINCT order_id) AS unique_orders,

COUNT(DISTINCT customer_id) AS unique_customers,

ROUND(SUM(sales), 2) AS total_sales,

ROUND(SUM(profit), 2) AS total_profit,

ROUND(SUM(profit) / SUM(sales)* 100, 2) AS profit_margin_percent,

ROUND(AVG(discount)* 100, 2),

COUNT(DISTINCT order_id) AS unique_orders,

COUNT(DISTINCT customer_id) AS unique_customers


FROM superstore;

""")

overall_performance
```

## Region Performance

```python
region_performance = run_query ("""

SELECT region, 

COUNT(DISTINCT order_id) AS unique_orders,

ROUND(SUM(sales), 2) AS total_sales,

ROUND(SUM(profit), 2) AS total_profit,

ROUND(SUM(profit) / SUM(sales)* 100, 2) AS profit_margin_percent

FROM superstore

GROUP BY region

ORDER BY total_profit DESC;

""")

region_performance
```

# Discount Impact 

### Discounting was analyzed to understand whether higher discounts are associated with weaker profitability. This is important because discounts may increase sales volume while reducing or eliminating profit.

```python
discount_performance = run_query ("""

SELECT discount_type, COUNT(*) AS no_orders,

COUNT(DISTINCT order_id) AS unique_orders,

ROUND(SUM(sales), 2) AS total_sales,

ROUND(SUM(profit), 2) AS total_profit,

ROUND(SUM(profit) / SUM(sales)* 100, 2) AS profit_margin_percent

FROM superstore

GROUP BY discount_type

ORDER BY profit_margin_percent DESC;

""")

discount_performance
```

```python
plt.figure(figsize=(8, 5))

plt.bar(discount_performance["discount_type"], discount_performance["total_profit"])

plt.title("Total Profit By Discount Type")

plt.xlabel("Discount Type")

plt.ylabel("Total Profit")

plt.xticks(rotation=0)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(discount_performance["discount_type"], discount_performance["profit_margin_percent"])

plt.title("Profit Margi By Discount Type")

plt.xlabel("Discount Type")

plt.ylabel("Profit Margin %")

plt.xticks(rotation=0)

plt.show()
```

```python
discount_grain = run_query ("""

SELECT discount, COUNT(*) AS no_orders,

COUNT(DISTINCT order_id) AS unique_orders,

ROUND(SUM(sales), 2) AS total_sales,

ROUND(SUM(profit), 2) AS total_profit,

ROUND(SUM(profit) / SUM(sales)* 100, 2) AS profit_margin_percent

FROM superstore

GROUP BY discount

ORDER BY discount;

""")

discount_grain
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(discount_grain["discount"], discount_grain["profit_margin_percent"])

plt.title("Profit Margi By Discount")

plt.xlabel("Discount Type")

plt.ylabel("Profit Margin (%)")

plt.show()
```

## Stop or significantly reduce discounts above 20%. The data shows a clear inverse relationship: as discounts increase beyond 20%, the profit_margin_percent drops sharply into negative territory. You are currently losing significant money on every order with a discount of 30% or higher. Focus marketing efforts on the 0%–20% discount range, where profit margins remain positive.

```python
category_discount_performance = run_query ("""

SELECT category, 
       discount_type, 
       COUNT(*) AS no_orders,
       COUNT(DISTINCT order_id) AS unique_orders,
       ROUND(SUM(sales), 2) AS total_sales,
       ROUND(SUM(profit), 2) AS total_profit,
       ROUND(SUM(profit) / SUM(sales)* 100, 2) AS profit_margin_percent

FROM superstore

GROUP BY category, discount_type

ORDER BY category, profit_margin_percent DESC;

""")

category_discount_performance
```

## Monthly Sales & Profit Trends

```python
monthly_trend = run_query ("""
SELECT
       order_month, 
       COUNT(DISTINCT order_id) AS unique_orders,
       ROUND(SUM(sales), 2) AS total_sales,
       ROUND(SUM(profit), 2) AS total_profit,
       ROUND(SUM(profit) / SUM(sales)* 100, 2) AS profit_margin_percent

FROM superstore

GROUP BY order_month 

ORDER BY order_month;

""")

monthly_trend.head(10)
```

```python
plt.figure(figsize=(14, 6))

plt.plot(monthly_trend["order_month"], monthly_trend["total_sales"], label="Sales")

plt.plot(monthly_trend["order_month"], monthly_trend["total_profit"], label="Profit")

plt.title("Monthly Sales & Profit Trend")

plt.xlabel("Month")

plt.ylabel("Amount")

plt.xticks(rotation=90)

plt.legend

plt.show()
```

```python
yearly_trend = run_query ("""
SELECT
       SUBSTR(order_date, 1, 4) AS order_year, 
       COUNT(DISTINCT order_id) AS unique_orders,
       ROUND(SUM(sales), 2) AS total_sales,
       ROUND(SUM(profit), 2) AS total_profit,
       ROUND(SUM(profit) / SUM(sales)* 100, 2) AS profit_margin_percent

FROM superstore

GROUP BY SUBSTR(order_date, 1, 4)

ORDER BY order_year ;

""")

yearly_trend
```

```python
plt.figure(figsize=(8, 5))

plt.plot(yearly_trend["order_year"], yearly_trend["total_sales"], marker=0, label="Total Sales")

plt.plot(yearly_trend["order_year"], yearly_trend["total_profit"], marker=0, label="Total Profit")

plt.title("yearly Sales & Profit Trend")

plt.xlabel("Month")

plt.ylabel("Amount")

plt.xticks(rotation=90)

plt.legend

plt.show()
```

## Sales and profit changed over time, with some months showing stronger sales while profit did not always increase at the same rate. This supports the idea that sales volume should not be evaluated by itself. Profit margin is also important because it shows whether the business is earning enough profit from its sales.

```python
high_sales_negative_profit = run_query ("""

SELECT product_name,
       category,
       sub_category,
       COUNT(DISTINCT order_id) AS total_orders,
       ROUND(SUM(sales), 2) AS total_sales,
       ROUND(SUM(profit), 2) AS total_profit,
       ROUND(SUM(profit) / SUM(sales)* 100, 2) AS profit_margin_percent

FROM superstore

GROUP BY product_name, category, sub_category

HAVING SUM(sales) > 10000 
       AND SUM(profit) < 0

ORDER BY total_sales Desc;

""")

high_sales_negative_profit
```

# Product-Level Findings

### The product-level analysis revealed that the Canon imageCLASS 2200 Advanced Copier was the top-performing product in both sales and profitability, generating approximately $61.6K in sales and 25.2K in profit.

### The poorest-performing product in terms of profit was the Cubify CubeX 3D Printer Double Head Print, which generated a loss of roughly $8.9K. Notably, several of the least profitable products belonged to the Machines and Tables sub-categories.

### These findings highlight the importance of evaluating products using multiple performance metrics—including sales, profit, and profit margin—rather than relying solely on sales revenue when making business decisions.
